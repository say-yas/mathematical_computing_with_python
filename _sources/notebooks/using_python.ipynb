# Using Python in Jupyter Notebooks

*A quick test page for the site.*

Most lectures in this course happens inside Jupyter notebooks. A notebook mixes
code, its output, and formatted text (Markdown) in one document that runs in
your browser. This page is a notebook itself. Below is how to get one
installed and running, and how to check that the install actually works.


## Installing Jupyter

Open a terminal (macOS/Linux) or Anaconda Prompt (Windows) and install
Jupyter into your environment.

**If you're using conda:**
```bash
conda activate your-environment-name
conda install -c conda-forge notebook
```

**If you're using plain pip:**
```bash
pip install notebook
```

Either way, this adds a `jupyter` command you can run from that environment.


## Confirm the install

After installing, check that the command is actually available:

```bash
$ which jupyter
/path/to/your/environment/bin/jupyter
```

If nothing prints, the install didn't land in the environment you're
currently using, which is the most common source of "command not found"
errors.


## Give your environment a kernel

If you created a dedicated environment for this course (recommended), Jupyter
needs a **kernel** so it knows how to run code using that environment
specifically.

```bash
conda activate your-environment-name
pip install ipykernel
python -m ipykernel install --user --name your-environment-name --display-name "Python (your-environment-name)"
```

Once this is done, "Python (your-environment-name)" will show up as a
selectable kernel inside Jupyter, and code you run in a notebook will use
that environment's packages.


## Launch the notebook server

From any folder you want to work in:

```bash
jupyter notebook
```

This opens a browser tab showing the files in that folder. From there you
can create a new notebook or open an existing `.ipynb` file, like the Class 1
notebook in this book.


## A quick sanity check

Run the cell below. If it executes without error and prints the message,
your notebook environment is working correctly.


````python
import sys
print("Jupyter is running Python", sys.version.split()[0])
print("This cell ran successfully.")
````

## One thing to remember

Cells in a notebook can be run out of order, and variables stick around
between cells once you've run them. If your results look wrong, try
**Restart Kernel and Run All Cells** from the Kernel menu; this reruns
everything top to bottom and is the fastest way to rule out stale state as
the cause of a bug.
