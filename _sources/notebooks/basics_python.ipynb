## Variables and Data Types

A **variable** is a name that refers to a value. In Python you don't declare a variable's type ahead of time and you don't need a special keyword to create one. You just assign a value to a name with `=`:

```python
name = value
```

Python is **dynamically typed**: a variable's type is determined by the value assigned to it, and it can change if you reassign the variable to a value of a different type.

````python
x = 10
print(x, type(x))
x = 3.14
print(x, type(x))
x = 'cd'
print(x, type(x))
x = True
print(x, type(x))
````

### Comments

Comments begin with the pound (`#`) symbol. Everything from the `#` to the end of the line is ignored by Python - they're notes for humans reading the code, not instructions for the interpreter.

````python
x = 1 # assign the variable x
print(x) # print the variable x
````

## Core Data Structures

Beyond single values like numbers and strings, Python has a handful of built-in **containers** for holding collections of values. The main ones are:

| Structure | Syntax | Ordered | Mutable | Example |
|---|---|---|---|---|
| `list`  | `[]` | yes | yes | `[1, 2, 3]` |
| `tuple` | `()` | yes | no  | `(1, 2, 3)` |
| `dict`  | `{key: value}` | yes (insertion order) | yes | `{"a": 1}` |
| `set`   | `{}` | no  | yes | `{1, 2, 3}` |

*Ordered* means the items keep the position you put them in, so you can index into them. *Mutable* means you can change the contents after creation (add, remove, or overwrite elements); immutable structures can't be changed once created.

### Lists

A Python list is specified using square brackets `[]`. Lists are ordered and mutable - you can add, remove, or change elements after the list is created.

````python
x = [1, 2, 3]
print(x)
````

You can access elements using square brackets `[]` as well. Python is **0-indexed**, so the first index is `x[0]`.

````python
print(x[0])
````

The length of a list (or other collection) can be obtained using `len()`.

````python
print(len(x))
````

You can grow a list with `.append()`, and use **slicing** (`start:stop`) to pull out a sub-section. Slicing also works on tuples and strings.

````python
x.append(4)
x.append(5)
x.append([7,8])
print("List:", x)
print("First element:", x[0])
print("Last element:", x[-1])
print("Elements 1 through 2:", x[1:3])
print("Reversed:", x[::-1])
````

### Tuples

A tuple looks like a list but uses parentheses `()`, and unlike a list, it's **immutable**: once created, you can't change, add, or remove its elements. Tuples are often used for small, fixed groupings of values, like coordinates.

````python
coordinates = (40.7128, -74.0060)
print("Tuple:", coordinates)
print("Latitude:", coordinates[0])
````

### Dictionaries

A `dict` (dictionary) stores **key-value pairs**, written as `{key: value, ...}`. Instead of looking things up by position (like a list), you look them up by key.

````python
student = {"name": "Amara", "score": 92, "passed": True}
print("Dict:", student)
print("Student score:", student["score"])
print("keys:", student.keys())
print("values:", student.values())
print("items:", student.items())
````

### Sets

A `set` is an unordered collection of **unique** values, written with curly braces `{}`. Duplicate values are automatically dropped.

````python
unique_grades = {88, 92, 79, 92, 88}
print("Set:", unique_grades)
````

## Functions

Functions let us package logic for reuse instead of copying and pasting code. They're defined using the `def` keyword, followed by the function name, arguments, and then a colon `:`. **Scope is determined by indentation** (4 spaces).

### General form

```python
def function_name(parameter1, parameter2, ...):
    """Optional docstring describing what the function does."""
    # body: do something with the parameters
    return result  # optional; omit if the function has no output
```

- `parameter1, parameter2, ...` are placeholders filled in each time the function is called.
- `return` sends a value back to wherever the function was called; a function with no `return` implicitly returns `None`.
- You can return values from a function using `return`.

````python
def square(x):
    # returns the square of x
    return x * x

square(2)
````

Functions can take multiple arguments and **keyword arguments** (arguments with a default value, given as `name=default` in the definition). They can also return multiple values at once, separated by commas.

````python
def f(x, y, a=2):
    # a is a keyword argument
    return x + y, a * x

w, z = f(1, 3, a=1)
print(w)
print(z)
````

## Strings

Strings hold text and can be specified with either single quotes `'` or double quotes `"`. You shouldn't mix which one you're using within the same string.

````python
a = 'hello'
b = "hello"
a == b
````

Multi-line strings are specified with triple quotes - either `'''` or `"""`.

````python
a = """A multi-
line string"""

print(a)
````

### printf-style formatting

Strings can be formatted using a syntax similar to C's printf, with `%` placeholders.

````python
s = "value = %d" % 1
s
````

````python
s = "name: %s, age: %d, weight: %.1f" % ("Bob", 20, 160.0)
s
````

### The `format` method

Strings can be formatted using the `.format()` method, which fills in curly braces `{}` with the arguments you pass it.

````python
x = 1
print("x = {}".format(x))
````

You can use the `format` arguments in multiple positions or in different orders by specifying the index in the braces:

````python
x = 2
print("{0} * {0} = {1}".format(x, square(x)))
````

You can also use keyword arguments in `format`, which can be very helpful in complicated situations.

````python
print("{name1} said \"hello {name2}!\", and {name2} said \"hello {name1}!\"".format(name1="Alice", name2="Bob"))
````

### f-strings

In Python 3.6, formatted string literals - or **f-strings** - were introduced. These give another option for formatting strings, and are easy to read: similar to using keyword arguments with `format`, but names are pulled directly from the surrounding code. F-strings are denoted with an `f` before the first quote, as in `f'...'`.

````python
name = "Alice"
s = f"Hello, {name}!"
print(s)
````

A good reference for format string syntax is [the documentation](https://docs.python.org/3/library/string.html#format-string-syntax).

## Logic and Control Flow

Control flow lets your code make decisions and run different blocks depending on a condition. Python uses `if`, `elif`, and `else`. The `True` and `False` keywords are used as boolean values, and you should use `and`, `or`, and `not` (rather than symbols like `&&`) for logical expressions.

### General form

```python
if condition1:
    # runs if condition1 is True
elif condition2:
    # runs if condition1 is False and condition2 is True
else:
    # runs if none of the above are True
```

`elif` and `else` are both optional, and you can chain as many `elif`s as you need.

````python
def sign(x):
    """
    returns the sign of a number.
    """
    if x < 0:
        return -1
    elif x > 0:
        return 1
    else:
        return 0

print(sign(2), sign(-3), sign(0))
````

````python
if True:
    print(True)

if False:
    print(False)
````

````python
if True or False:
    print(True or False)
````

````python
if not False:
    print("hello")
````

## Iteration

Iteration means running the same block of code repeatedly - once per item in a collection, or until some condition changes. Python has two loop keywords: `for` and `while`. As with `if`, **indentation determines scope**.

### General forms

**`for` loop** - runs once per item in an iterable (a list, string, range, etc.):
```python
for item in iterable:
    # runs once per item in iterable
```

**`while` loop** - runs repeatedly as long as a condition stays `True`:
```python
while condition:
    # runs repeatedly as long as condition is True
    # something inside the loop must eventually make condition False
```

Inside a loop, `break` will exit the loop immediately, and `continue` will skip straight to the next iteration.

````python
for i in range(5):
    print(i)
````

````python
n = 0
while True:
    if n == 5:
        break
    print(n)
    n = n + 1
````

You can also iterate directly over the elements of a collection, such as a list, using `in`.

````python
x = [1, 2, 'a']
for i in x:
    print(i)
````

````python
scores = {"Amara": 92, "Ben": 78, "Cleo": 65}
for name, score in scores.items():
    if score >= 90:
        grade = "A"
    elif score >= 80:
        grade = "B"
    elif score >= 70:
        grade = "C"
    else:
        grade = "F"
    print(name, grade)
````
