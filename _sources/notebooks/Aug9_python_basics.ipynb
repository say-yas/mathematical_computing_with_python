````python
# Basic arithmetic
a, b = 7, 2

print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)   # true division -> float
print("a // b =", a // b) # floor division -> int
print("a % b =", a % b)   # remainder
print("a ** b =", a ** b) # exponent
````

## 2. Core Data Structures

| Structure | Ordered | Mutable | Example |
|---|---|---|---|
| `list`  | yes | yes | `[1, 2, 3]` |
| `tuple` | yes | no  | `(1, 2, 3)` |
| `dict`  | yes (insertion order) | yes | `{"a": 1}` |
| `set`   | no  | yes | `{1, 2, 3}` |


````python
scores_list = [88, 92, 79, 95, 61, 74]   # list: ordered, mutable
scores_list.append(85)
print("List:", scores_list)

coordinates = (40.7128, -74.0060)        # tuple: ordered, immutable
print("Tuple:", coordinates)

student = {"name": "Amara", "score": 92, "passed": True}  # dict
print("Dict:", student)
print("Student score:", student["score"])

unique_grades = {88, 92, 79, 92, 88}     # set: drops duplicates
print("Set:", unique_grades)
````

````python
# Slicing works on lists, tuples, and strings
print(scores_list[0])     # first element
print(scores_list[-1])    # last element
print(scores_list[1:4])   # elements at index 1, 2, 3
print(scores_list[::-1])  # reversed
````

## 3. Control Flow: `if`, `for`, `while`


````python
score = 74

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print(f"A score of {score} earns a grade of {grade}")
````

````python
# for loop over a list
for s in scores_list:
    status = "pass" if s >= 70 else "fail"
    print(f"score={s:>3}  ->  {status}")
````

````python
# while loop
count = 0
total = 0
while count < len(scores_list):
    total += scores_list[count]
    count += 1

print("Total:", total)
print("Average:", total / len(scores_list))
````

## 4. Functions

Functions let us package logic for reuse. This is essential once our mathematical
computations get more involved.


````python
def average(numbers):
    """Return the arithmetic mean of a list of numbers."""
    return sum(numbers) / len(numbers)

def letter_grade(score):
    """Convert a numeric score to a letter grade."""
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    else:
        return "F"

print("Average score:", average(scores_list))
print("Letter grades:", [letter_grade(s) for s in scores_list])  # list comprehension
````

### A quick note on list comprehensions
`[letter_grade(s) for s in scores_list]` is shorthand for:
```python
result = []
for s in scores_list:
    result.append(letter_grade(s))
```
You'll see this compact style constantly in mathematical/scientific Python code.


## 5. Introducing NumPy

Plain Python lists are flexible but slow for numerical work, and they don't
support element-wise math out of the box. **NumPy** gives us the `ndarray`
object: a fast, fixed-type, multi-dimensional array with vectorized
operations.

Let's rebuild our gradebook example as a small **dataset** and explore it
with NumPy.


````python
import numpy as np

np.random.seed(42)  # reproducibility: everyone gets the same "random" data

# Simulate a small dataset: exam scores for 30 students across 3 exams
n_students = 30
exam1 = np.random.normal(loc=75, scale=10, size=n_students).round(1)
exam2 = np.random.normal(loc=78, scale=8, size=n_students).round(1)
exam3 = np.random.normal(loc=82, scale=9, size=n_students).round(1)

# Clip so nobody scores above 100 or below 0
exam1, exam2, exam3 = (np.clip(e, 0, 100) for e in (exam1, exam2, exam3))

# Stack into one 2D array: rows = students, columns = exams
gradebook = np.column_stack([exam1, exam2, exam3])

print("Shape (students, exams):", gradebook.shape)
print(gradebook[:5])  # first 5 students
````

Notice `gradebook` is a single NumPy array containing everyone's scores.
Unlike the plain Python list from Section 2, this is a genuine **dataset**:
rows are observations (students), columns are variables (exams) — a layout
you'll see constantly in mathematical computing and data analysis.


````python
# Indexing and slicing a 2D array
print("First student's scores:", gradebook[0])
print("Everyone's Exam 2 scores:", gradebook[:, 1])
print("First 3 students, all exams:")
print(gradebook[:3, :])
````

````python
# Vectorized operations: no explicit loop needed!
final_scores = gradebook.mean(axis=1)   # average across exams, per student
print("Final average per student (first 5):", final_scores[:5].round(1))

class_average_per_exam = gradebook.mean(axis=0)  # average per exam, across students
print("Class average per exam:", class_average_per_exam.round(1))
````

````python
# Descriptive statistics
print("Overall mean:   ", gradebook.mean().round(2))
print("Overall std dev:", gradebook.std().round(2))
print("Min score:      ", gradebook.min())
print("Max score:      ", gradebook.max())
print("Median (Exam 1):", np.median(exam1).round(2))
````

````python
# Boolean masking: a core NumPy idiom
passing_mask = final_scores >= 70
print("Number of students passing overall:", passing_mask.sum())
print("Passing students' final scores:", final_scores[passing_mask].round(1))
````
